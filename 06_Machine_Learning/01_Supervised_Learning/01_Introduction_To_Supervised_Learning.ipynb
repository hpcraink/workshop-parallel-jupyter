# Introduction to Supervised Learning

**Machine learning** algorithms use statistics to find patterns in large amounts of
data, and then use those patterns to make predictions. Assessing creditworthiness,
forecasting prices, filtering spam — all of them work the same way: show the
algorithm enough past examples, and it works out the rule for itself.

The quality of those predictions depends above all on the quality of the data.

This chapter is split in two, and the split follows the most important distinction
in machine learning:

- **`01_Supervised_Learning`** — this folder. The past data comes with the answers
  attached, and the algorithm learns to reproduce them.
- **`02_Unsupervised_Learning`** — the other folder. There are no answers, and the
  algorithm has to find structure in the data on its own.

This notebook introduces the vocabulary and the workflow that the rest of this
folder relies on. We deliberately use a tiny, artificial dataset here, so that
nothing distracts from the four steps every supervised model goes through.

## 1. Supervised and unsupervised learning

**Supervised learning** works from past data where *the desired output is already
known*. For a set of taxi rides we know the hour, the number of passengers, the
distance — and we also know what each ride actually cost. Data like this is called
**labelled**, and the label is the answer we want the model to reproduce for rides
it has never seen.

Supervised problems come in two kinds, and the difference is simply what sort of
answer you are after:

| Kind | The label is | Example |
|---|---|---|
| **Regression** | a continuous number | the fare of a taxi ride, the price of a house |
| **Classification** | one of a few fixed categories | handwriting recognition, spam or not spam |

**Unsupervised learning** is the case where no labels exist at all. Nobody has said
what the right answer is, so the algorithm has to find structure by itself — for
instance by discovering that the data falls into groups. That is the subject of the
`02_Unsupervised_Learning` folder; everything here is supervised.

<a id="xy"></a>
## 2. Features and labels

Two letters appear in essentially every machine learning example, including all the
notebooks in this chapter:

- **`X`** — the **features**: the data used to make the estimate
- **`y`** — the **label**: the value we want to predict

For taxi rides, `X` would hold the hour, the passenger count and the distance, while
`y` holds the fare.

<a id="split"></a>
## 3. Training data and test data

A model that has memorised its examples will answer them perfectly and still be
useless. To find out whether it has learned anything that *generalises*, part of the
data is held back: the model is trained on one part and checked against the other,
which it has never seen.

Say we have nine taxi rides. Six of them are given to the model to learn from — it
sees the three features *and* the fare that each ride actually cost:

**Training data**

| Hour | # Passengers | Distance travelled | Fare (USD) |
|------|--------------|--------------------|-----------:|
| 23   | 5            | 10                 | 10         |
| 22   | 3            | 14                 | 11         |
| 20   | 2            | 17                 | 15         |
| 18   | 1            | 5                  | 8          |
| 13   | 1            | 4                  | 6          |
| 15   | 4            | 43                 | 43         |

The other three are kept back. The model is shown only the first three columns and
has to produce the fourth itself. We know what it should be — that is how we can
grade the answer — but the model never saw these rides while learning:

**Test data**

| Hour | # Passengers | Distance travelled | Fare (USD) |
|------|--------------|--------------------|-----------:|
| 16   | 1            | 22                 | 27         |
| 12   | 2            | 1                  | 3          |
| 09   | 3            | 8                  | 8          |

That gives the four names used throughout: **`X_train`** and **`y_train`** to learn
from, **`X_test`** and **`y_test`** to be judged on.

## 4. scikit-learn

**scikit-learn** is the standard Python library for classical machine learning. Its
main virtue is uniformity: every algorithm it offers is used through the same few
method calls, so swapping one model for another is a one-line change. It also brings
the tools for splitting data, validating models and comparing them, and it is
exceptionally well documented.

That uniform interface is what the rest of this notebook is really about.

<a id="workflow"></a>
## 5. The workflow, step by step

Every supervised model in scikit-learn follows the same four steps:

1. **split** the data into training and test parts
2. **fit** the model on the training part
3. **predict** the labels of the test part
4. **score** those predictions against the labels we kept back

We walk through them on a deliberately tiny dataset — six rows, two features — so
that every intermediate result fits on screen.

```python
import numpy as np
from sklearn.model_selection import train_test_split

X = np.arange(12).reshape((6, 2))    # 6 rows, 2 features
y = np.arange(6)                     # the label of each row

X
```

```python
y
```

Be aware of what this data is: `y` is simply the row number, and `X` counts upwards
in step with it. The relationship is exact, so any model will find it immediately.
That is deliberate — the point here is the mechanics, not the difficulty.

`train_test_split` shuffles the rows and cuts them in two. Without arguments it
keeps 25% for testing.

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=0)

print("training rows:", len(X_train))
print("test rows:    ", len(X_test))
```

Four rows to learn from, two to be tested on. The proportion is adjustable —
`test_size=0.5` splits the six rows evenly.

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.5, random_state=0)

print("training rows:", len(X_train))
print("test rows:    ", len(X_test))
```

Now the model itself. Importing it, creating it and training it are three lines —
and the same three lines, with a different name on the first, would train any other
scikit-learn model.

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(X_train, y_train)
```

`predict` applies the trained model to data it has not seen.

```python
predictions = model.predict(X_test)

predictions
```

And `mean_squared_error` compares those predictions with the labels we held back.
It averages the squared difference between them, so **0 means perfect** and larger
is worse.

```python
from sklearn.metrics import mean_squared_error

print("true labels:", y_test)
print("predicted:  ", predictions.round(3))
print()
print("mean squared error:", mean_squared_error(y_test, predictions))
```

Essentially zero — but as warned above, that is a property of the data rather than
an achievement of the model. `y` was the row index and `X` counted along with it, so
there was a perfect straight line to find.

Real data never looks like this. [Notebook 02](02_Linear_Regression.ipynb) runs the
same four steps on 1000 real taxi rides, and the result is far less flattering.

## 6. Why the amount of training data matters

One more thing this toy example can show. Here is the identical procedure on a
dataset of **two rows** — after the split, the model has exactly one example to
learn from.

```python
X = np.arange(4).reshape((2, 2))
y = np.arange(2)

X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=0)

print("training rows:", len(X_train))
print("test rows:    ", len(X_test))
```

```python
model = LinearRegression()
model.fit(X_train, y_train)

predictions = model.predict(X_test)

print("true label:", y_test)
print("predicted: ", predictions.round(3))
print()
print("mean squared error:", mean_squared_error(y_test, predictions))
```

The error jumps from almost 0 to **1.0**, on data with exactly the same structure.
Nothing about the problem changed — only the number of examples the model was
allowed to see.

This is the single most reliable lesson in machine learning: *more and better data
beats almost any amount of cleverness in the choice of model.*
