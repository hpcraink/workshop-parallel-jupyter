## hvPlot

hvPlot is a high level-API for data visualization. This means it hands the user an easy way to create different types of plots and is relatively easy to use with only few lines of code. <br>
hvPlot itself is able to use different plotting-backends like matplotlib or Plotly to generate it's plots, so you can decide whether you want an interactive plot from Plotly or a static object from matplotlib. (You could also create static plots with Plotly or (somewhat) interactive plots with matplotlib, but their respective defaults are interactive for Plotly and static for matplotlib)

hvPlot is able to comprehend data inputs from, amongst others, pandas and xarray (xarray is basically a fancy combination of pandas and numpy).

Further information on hvplot can be found on their website https://hvplot.holoviz.org/

So, to plot the information of our pandas Dataframe, we have to import the pandas functionality of hvplot.

```python
import hvplot.pandas
```

```python
df_sorted_fromfile.hvplot.bar(x="Name", y="Population")
```

The default plotting backend of hvplot is Bokeh. So if we want to use another backend (and thus change the plot's behaviour), we have to change it with a command.

```python
# The same plot from a different backend (Plotly)
hvplot.extension('plotly')

df_sorted_fromfile.hvplot.bar(x="Name", y="Population")
```

```python
# And from another backend (matplotlib)
hvplot.extension('matplotlib')

df_sorted_fromfile.hvplot.bar(x="Name", y="Population")
```

Now that we changed the plotting backend, every plot we generate from here on out, will use the backend that was defined last! So if we want to use the Bokeh backend again, we'd have to switch back to it.

```python
df_sorted_fromfile.hvplot.line(x="Name", y="Population")
```

```python
# Back to the Bokeh backend
hvplot.extension('bokeh')

df_sorted_fromfile.hvplot.line(x="Name", y="Population")
```

## Matplotlib
Matplotlib is the most widely used data visualization library in python.
It's a bit older and can sometimes feel a bit rusty compared to it's newer alternatives, but is still very powerful and a favorite for many python coders. You can change a lot of properties of your plots. But to do so, you have to give lots of input to matplotlib!

### Basic Plotting

```python
# Let's import the pyplot interface to matplotlib
import matplotlib.pyplot as plt

# And tell matplotlib to plot inline within the notebook
%matplotlib inline
```

Let's start with a very simple line-plot (`plt.plot()`) using our pandas dataframe:

```python
plt.plot(df["Name"], df["Population"])

plt.show()
```

When creating a visualization with matplotlib, you can adjust most parameters of the plot.

And as you can see from the example above, you really should do that if you want a nice result.

Let's start with some basic adjustments:

```python
# Creates a lineplot object with a label
plt.plot(df["Name"], df["Population"], linewidth=5, label="Our Data")

# Adds a title to the plot object
plt.title("University Cities in Baden-Württemberg:\n Total Population", fontsize=15)

# Adding labels to the axes
plt.xlabel("City", labelpad=10, fontsize=15)
plt.ylabel("Population", labelpad=10, fontsize=15)

# Adds a legend to the plot. As no arguments were given, placement is done
# automated by matplotlib.
plt.legend()

# Editing the properties of the tick markers
plt.xticks(rotation=60)

plt.show()
```

Much better!

We can easily change the type of plot:

```python
plt.bar(df["Name"], df["Population"],  lw=5)

plt.title("University Cities in Baden-Württemberg:\n Total Population", fontsize=15)

plt.xlabel("City", labelpad=10, fontsize=15)
plt.ylabel("Population", labelpad=10, fontsize=15)

plt.xticks(rotation=60)

plt.show()
```

### Scatter Plots

In this following section, we will focus on so-called scatter plots.

Let us now generate a set of two-dimensional data points, simulating two populations in front of a uniform background.

```python
# Uniform background data
x1 = np.random.uniform(0,10,1000)
y1 = np.random.uniform(0,10,1000)
# First population
x2 = np.random.normal(6,1,1000)
y2 = np.random.normal(5,1,1000)
# second population
x3 = np.random.normal(2,.5,1000)
y3 = np.random.normal(4,1,1000)
```

We have generated background data using `np.random.uniform()` as well as two distributions by sampling from a normal distribution with `np.random.normal()`.

Let's have a look at it:

```python
plt.scatter(x1, y1,  lw=1)
plt.scatter(x2, y2,  lw=1)
plt.scatter(x3, y3,  lw=1)

plt.show()
```

We can change the opacity using the `alpha` parameter:

```python
plt.scatter(x1, y1,  lw=1, alpha=0.2)
plt.scatter(x2, y2,  lw=1, alpha=0.2)
plt.scatter(x3, y3,  lw=1, alpha=0.2)

plt.show()
```

Lets now join our three distributions into one numpy array using `np.concatenate`data, and convert them to a pandas dataframe:

```python
data = np.vstack([np.concatenate([x1,x2,x3]), np.concatenate([y1,y2,y3])]) # there are many ways how to combine numpy arrays. Here we combine np.vstack and np.concatenate.
data_df = pd.DataFrame(data.T, columns=['x', 'y'])

plt.scatter(data_df["x"], data_df["y"], alpha=0.2)

plt.show()
```

```python
data_df = pd.DataFrame(data.T, columns=['x', 'y']) # data.T: we want our data organized in columns for a DataFrame, so we transpose the numpy arry
data_df
```

We can now combine the scatter- and histogram plots into a single figure:

```python
import matplotlib.gridspec as gridspec 

fig_sc = plt.figure(figsize=(6,6))

gs = gridspec.GridSpec(3, 3)
ax_scatter = plt.subplot(gs[1:3, :2])
ax_histx = plt.subplot(gs[0, :2], sharex=ax_scatter)
ax_histy = plt.subplot(gs[1:3, 2], sharey=ax_scatter)
    
ax_scatter.scatter(data_df["x"], data_df["y"], alpha=0.2, lw=1)
ax_scatter.set(xlabel="x data", ylabel="y data")

ax_histx.hist(data_df["x"], bins=100, align='mid')
ax_histx.set(ylabel='Count')

ax_histy.hist(data_df["y"],bins=100, orientation='horizontal', align='mid')
ax_histy.set(xlabel='Count')

fig_sc.show()
```

We can also generate a two-dimensional histogram of our data:

```python
fig, axs = plt.subplots(nrows=2, ncols=2, figsize=(6,6))

# Create a scatter plot in the 0,0 subplot (top left)
axs[0, 0].set_title('Scatter Plot')
axs[0, 0].scatter(data_df["x"], data_df["y"], alpha=0.2, lw=1)

# Create a 2D-Hisogram/heatmap in the 0,1 subplot (top right)
axs[0, 1].set_title('2D-Histogram')
axs[0, 1].hist2d(data_df["x"], data_df["y"], bins=20, cmap='viridis') # here we use matplotlibs's 2d histogram function. 

# Create a 2D-Hisogram with hexagonal bins in the 1,0 subplot (bottom left)
axs[1,0].set_title('Hexbin Plot')
axs[1,0].hexbin(data_df["x"], data_df["y"], gridsize=20, cmap='viridis') 

# Create a contour plot in the 1,1 subplot (bottom right)
axs[1,1].set_title('Contour Plot')
# Here we use numpy's 2d histogram function.
hist, xedges, yedges = np.histogram2d(data_df["y"], data_df["x"], density=True, bins=20)

# This is plotting the contour lines
CL = axs[1,1].contour(hist, extent=[xedges.min(),xedges.max(),yedges.min(),yedges.max()],
                      levels=10, colors='silver',
                      linestyles ='--', linewidths=0.5) 
# This fills the contour levels
axs[1,1].contourf(hist, extent=[xedges.min(),xedges.max(),yedges.min(),yedges.max()], 
                  levels=20, cmap='viridis')

# This for loop will set the plot boundaries for all subplots. Otherwise matplotlib
# would add whitespace around the actual plotted data.
for x in range(2):
    for y in range(2):
        axs[x, y].set_xlim([0,10])
        axs[x, y].set_ylim([0,10])
fig.tight_layout()

fig.show()
```

As you can see, it is basically impossible to determine the difference between sparsely popoulated and empty bins in the 2D-hisogram and the hexbin plot. To make that difference clear, one can define a minimum value to be plotted with the vmin keyword. This value has to be very small, yet still larger than 0, so only truly empty bins won't be plotted.
Also one has to set a background colour to be displayed instead.

```python
# Define the colormap
vir = plt.get_cmap("viridis")
# Define white as background color
vir.set_under('w')

# Now the same with the plasma colormap
plasm = plt.get_cmap("plasma")
plasm.set_under('w')
```

```python
fig, axs = plt.subplots(nrows=2, ncols=2, figsize=(6,6))

axs[0, 0].set_title('Scatter Plot')
axs[0, 0].scatter(data_df["x"], data_df["y"], alpha=0.2, lw=1)

axs[0, 1].set_title('2D-Histogram')
# Now we add the vmin statement and our adjusted colormap.
axs[0, 1].hist2d(data_df["x"], data_df["y"], bins=20, cmap=vir, vmin=1e-14) 

axs[1,0].set_title('Hexbin Plot')
axs[1,0].hexbin(data_df["x"], data_df["y"], gridsize=20, cmap=plasm, vmin=1e-14) 

axs[1,1].set_title('Contour Plot')
hist, xedges, yedges = np.histogram2d(data_df["y"], data_df["x"], density=True, bins=20) # here we use numpy's 2d histogram function. 
CL = axs[1,1].contour(hist, extent=[xedges.min(),xedges.max(),yedges.min(),yedges.max()],
                      levels=10, colors='silver', linestyles ='--', linewidths=0.5)
axs[1,1].contourf(hist, extent=[xedges.min(),xedges.max(),yedges.min(),yedges.max()],
                  levels=20, cmap="viridis")

for x in range(2):
    for y in range(2):
        axs[x, y].set_xlim([0,10])
        axs[x, y].set_ylim([0,10])
fig.tight_layout()

fig.show()
```

We can also add a new trace to our figure-object. To do this, let's first create some new data.

```python
# Create a 2-D normal distribution for 1000 points with x-location x=8 and x-stdev 1, and y-location y=6.5 and y-stdev 0.5

# You can set your pointer to the end of np.random.normal or within the function's accompaning braces()
# and press Shift+Tab to get an info field, which more closely explains the function.
x_distribution = np.random.normal(8, 0.5, 100)
y_distribution = np.random.normal(6.5, 0.25, 100)

data = np.vstack([x_distribution, y_distribution])
new_data_df = pd.DataFrame(data.T, columns=['x', 'y'])
```

```python
Figure_two.add_scatter(x=new_data_df["x"],
                       y=new_data_df["y"],
                       marker_symbol="star",
                       marker_color="crimson",
                       marker_opacity=0.4,
                       # Name of the new trace:
                       name="new_data",
                       # add_scatter is used to add line and or scatter plots, so we tell it to
                       # only use markers.
                       mode="markers",
                      )

Figure_two.show()
```

## Scientific Graphic Design

### First steps
The first step in scientific graph/plot design is always to create a graphic, that represents your data properly. In this stage it might be useful to experiment with different types of plots or different plot properties (e.g. logarithmic scales) to help explore the data. <br>
At this stage, small details of the graphics (figure size, text size, positions, axis notations) do not have to be defined exactly. Most important is, that the data is represented well and that people familiar with the project can interpret the graphic.

### Interconnecting and refining
Once the data is explored and your goal is communication with people less familiar with your work, you should consider how to present your data. Some important points to consider typically are:

- What data is relevant to the topic?
- Which plots/graphics complement each other / belong together?
- Are there redundancies? Are there good reasons to keep them?
- How much explaining is needed to understand the graphic?
- Do axis titles and ticks clarify what is displayed?
- What knowledge is needed to understand/interpret the graphics?
- Would some other graphic help with understanding?
- Could some color-coding assist comprehension?

### Settling and polishing
At some point, you will have to think about publication/presentation of your research. For this, you will have to tackle further issues with your plots and graphics:

- How much space is available to me?
- Graphics and plots have to be set at a specific size.
- Are titles and ticks easy to read at the final size?
- Are graphs grouped together in a sensible way?
- Is the color coding consistent throughout? Can I make it consistent?
- Can my graphs be perceived correctly by everyone (exception: severe sight impediment such as blindness)?

The last bullet point includes considerations regarding different forms of colorblindness and/or prints in grey-scales. For continous colorbars usually the colorbar serves as a third axis for your data, so it should be treated as such. This means it best has a clear and constant gradient (usually within the lightness).
